# Numpy Array Operations:

```python
import numpy as np
```

```python
# 1.creating 1D NumPy array 

temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])  
temperatures_w1
```

```python
# 2. Inspection and properties
# inspecting the shape 

temperatures_w1.shape  
```

```python
# inspecting the data type

temperatures_w1.dtype   
```

```python
 # inspecting the number of elements

temperatures_w1.size   
```

```python
# inspecting the dimension of the given array

temperatures_w1.ndim
```

```python
# 3. Array operations 
# converting the temperatures_w1 from Celsius to Fahrenheit
F_temperatures_w1=np.multiply(temperatures_w1,9/5)+32
F_temperatures_w1
```

```python
# maximum temperature of the week
Max=np.max(F_temperatures_w1)
Max
```

```python
# minimum temperature of the week
Min=np.min(F_temperatures_w1)
Min
```

```python
# mean of the temperatures of the week
Mean=np.mean(F_temperatures_w1)
Mean
```

```python
# 4. Array Slicing and Indexing

print(F_temperatures_w1)
F_temperatures_w1[:3]  # The temperatures for the first three days of the week.
```

```python
print(F_temperatures_w1)
F_temperatures_w1[5:]   # The temperatures for the weekend (last two days).
```

```python
print(F_temperatures_w1)
F_temperatures_w1[2:5] # The temperatures for the middle three days of the week.
```

```python
# 5. Creating a 2D array

temperatures=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
temperatures
```

```python
# 6. Inspecting and slicing 

temperatures.ndim  # Inspecting the dimension of the given array
```

```python
temperatures.shape  # Inspecting the shape
```

```python
temperatures.dtype # inspecting the data type
```

```python
temperatures.size  # inspecting the number of elements
```

```python
print(temperatures)
temperatures[0:] # extracting the temperatures of both week 1 and week 2
```

```python
print(temperatures)
temperatures[0:,-2:] # extracting the temperatures of the weekends for both week 1 and week 2
```

# Pandas Series and DataFrame Operations

## Pandas Series:

```python
import pandas as pd
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
# 1. creating pandas Series
Marks=[95, 92, 89, 85, 80]
Custom_Indices=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
MARKS=pd.Series(index=Custom_Indices,data=Marks,name='MARKS')
MARKS
```

```python
# 2. Indexing and Slicing 
MARKS.iloc[0]    # To access the mark of 1st rank student
```

```python
MARKS.loc[['Rank1','Rank2','Rank3']]  # Retrieving the marks of the top 3 ranks by specifying their index labels.
```

```python
MARKS.iloc[2]   # Retrieving the mark of the 3rd rank student.
```

```python
MARKS[MARKS>90]  # Applying a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
```

```python
# 3. Manipulating Series
MARKS.loc['Rank1']=100     # Modifying the mark of the student with the 1st rank to 100.
MARKS 
```

```python
MARKS.drop('Rank5',inplace=True)  # Removing the entry corresponding to the last rank from the Series.
MARKS
```

```python
CGPA= (MARKS/10)    # Computing the CGPA by dividing each mark in the Series by 10.
CGPA
```

## Pandas DataFrame:

```python
# 1. creating pandas DataFrame
data ={
    'Transaction_ID':[101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Product_Category':['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture',
                   'Electronics'],
    'Region':['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount':[200, 150, 300, 450, 200, 250, 300, 180, 350, 400] }

transactions=pd.DataFrame(data)
transactions
```

```python
# 2. Data Exploration
# Display of DataFrame and its basic information, including head, tail, shape, column names, and data types.
print(transactions)
print()
print(transactions.head())
print()
print(transactions.tail())
print()
print(transactions.shape)
print()
print(transactions.columns)
print()
print(transactions.dtypes)
print()
```

```python
transactions[['Product_Category','Amount']] # Displaying columns
```

```python
transactions.loc[0:][['Product_Category','Region','Amount']]   # Retrieving the last 3 columns using loc method
```

```python
transactions[(transactions['Region']=='North') & (transactions['Amount']>200)]

# Filtering rows where the 'Region' is 'North' and 'Amount' is greater than 200.
```

```python
transactions['Product_Category'].value_counts()   # find the value_counts for the 'Product_Category' column
```

```python
transactions['Region'].unique()   # find the unique values in the 'Region' column
```

```python
transactions.groupby('Region')['Amount'].mean()    # Group by 'Region' and find the mean amount for each region.
```

```python
# 3. Manipulating the DataFrame
transactions.loc[transactions['Transaction_ID']==102, 'Amount']=165
transactions
```

```python
transactions['Discount']=transactions['Amount']*0.10
transactions
```

```python
transactions=transactions[transactions['Transaction_ID'] !=109]
transactions    
```

```python
transactions=transactions.drop('Discount',axis=1)
transactions
```
